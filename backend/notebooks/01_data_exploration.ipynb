```python
# Import Pandas
import pandas as pd

# Import NumPy
import numpy as np

# Import Matplotlib
import matplotlib.pyplot as plt

# Import Seaborn
import seaborn as sns

# Display all columns when showing DataFrames
pd.set_option("display.max_columns", None)

# Display all rows when needed
pd.set_option("display.max_rows", 100)
```

```python
# Load data from CSV file
df = pd.read_csv("../data/heart.csv")

# Display the first 5 rows of the DataFrame
print(df.head())
```

```python
# shape
print("Shape of the DataFrame:", df.shape)
```

```python
# Display all the columns in the DataFrame
print("Columns in the DataFrame:", df.columns.tolist())
```

```python
# display dataset information
print("Dataset Information:")
print(df.info())
```

```python
# describe statistical summary of the dataset
print("Statistical Summary of the Dataset:")
print(df.describe())
```

```python
# Count the number of missing values in each column
missing_values = df.isnull().sum()
print("Missing Values in Each Column:")
print(missing_values)
```

```python
# Count duplicate rows
df.duplicated().sum()
```

```python
# Count the number of unique values in each column
print("Number of Unique Values in Each Column:")
print(df.nunique())
```

```python
print("Value Counts for Target Variable:")
print(df.target.value_counts())
```

```python
# calculate target class percentage
target_class_percentage = df.target.value_counts(normalize=True) * 100
print("Target Class Percentage:")
print(target_class_percentage)
```

```python
# Visualize target distribution

# Set chart size
plt.figure(figsize=(7, 5))

# Create count plot
sns.countplot(data=df, x="target")

# Add chart title
plt.title("Heart Disease Target Distribution")

# Add axis labels
plt.xlabel("Heart Disease (0 = Absence, 1 = Presence)")
plt.ylabel("Number of Patients")

# Display chart
plt.show()
```

### Feature Distribution Analysis

```python
# Select numerical columns
numerical_features = [
    "age",
    "trestbps",
    "chol",
    "thalach",
    "oldpeak"
]

# Generate statistical summaries
df[numerical_features].describe()
```

```python
# Visualize numerical features using histograms
# Create a figure with multiple subplots
plt.figure(figsize=(14, 10))

# Generate histograms for numerical features
for index, feature in enumerate(numerical_features, 1):

    # Create subplot
    plt.subplot(3, 2, index)

    # Draw histogram with density curve
    sns.histplot(data=df, x=feature, kde=True)

    # Add title
    plt.title(f"Distribution of {feature}")

# Adjust spacing
plt.tight_layout()

# Display plots
plt.show()
```

### Categorical Feature Analysis

```python
# Define categorical features
categorical_features = [
    "sex",
    "cp",
    "fbs",
    "restecg",
    "exang",
    "slope",
    "ca",
    "thal"
]

# Create figure
plt.figure(figsize=(16, 12))

# Generate count plots
for index, feature in enumerate(categorical_features, 1):

    # Create subplot
    plt.subplot(4, 2, index)

    # Create count plot
    sns.countplot(data=df, x=feature)

    # Add title
    plt.title(f"Distribution of {feature}")

# Adjust layout
plt.tight_layout()

# Display plots
plt.show()
```

### Correlation Analysis

```python
# Calculate correlation between numerical columns
correlation_matrix = df.corr()

# Display correlation values
correlation_matrix
```

```python
# Visualize correlation using a heatmap

# Set figure size
plt.figure(figsize=(14, 10))

# Create correlation heatmap
sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5
)

# Add title
plt.title("Correlation Heatmap - Heart Disease Dataset")

# Display chart
plt.show()
```

```python
# Identify features correlated with target
# Extract correlations with target
target_correlation = correlation_matrix["target"]

# Sort by absolute correlation strength
target_correlation = target_correlation.abs().sort_values(
    ascending=False
)

# Display sorted correlations
target_correlation
```
